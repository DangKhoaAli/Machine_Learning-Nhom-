# **Descriptive Statistics Diabetes**

## Import các thư viện cần thiết

```python
import pandas as pd
import numpy as np
```

## Load dataset

```python
names = ['preg', 'glucose', 'pres', 'skin', 'insulin', 'mass', 'pedi', 'age', 'class']
df = pd.read_csv('pima-indians-diabetes.data.csv', header=None, names=names)
df
```

## Thông tin tổng quan về dữ liệu

- Số dòng: 768
- Số cột: 9
- Biến mục tiêu: `class` (0: không tiểu đường, 1: có tiểu đường)
- Các biến đặc trưng:
  - `Pregnancies(preg)`: Số lần mang thai
  - `Glucose`: Mức đường huyết
  - `BloodPressure (pres)`: Huyết áp
  - `SkinThickness (skin)`: Độ dày da
  - `Insulin`: Mức insulin
  - `BMI (mass)`: Chỉ số khối cơ thể
  - `DiabetesPedigreeFunction (pedi)`: Chỉ số di truyền tiểu đường
  - `Age`: Tuổi


## Kiểu dữ liệu của các thuộc tính trong DataFrame

```python
df.info()
```

Kết quả cho thấy:

- Dữ liệu gồm 768 dòng và 9 cột
- Không có cột nào có giá trị NaN
- Tất cả các cột đều có kiểu dữ liệu số (`int64` hoặc `float64`)

→ Tiếp theo, chúng ta sẽ kiểm tra thống kê mô tả để hiểu phân bố dữ liệu.


## Thống kê mô tả dữ liệu

```python
df.describe()
```

Phương thức này cho ta một cái nhìn tổng quan về dữ liệu trong dataset như:<br>
&#8226;&nbsp;&nbsp; `Count` : Đếm số lượng dòng có thông tin của 1 cột.<br>
&#8226;&nbsp;&nbsp; `mean` : Giá trị trung bình của 1 cột.<br>
&#8226;&nbsp;&nbsp; `std` : Độ lệch chuẩn của cột đó.<br>
&#8226;&nbsp;&nbsp; `min` : Giá trị nhỏ nhất của cột.<br>
&#8226;&nbsp;&nbsp; `25%, 50%, 75%` : Đây là tứ phân vị (quartiles) của cột, tìm giá trị nhỏ hơn hoặc bằng ở từng phân vị.<br>
&#8226;&nbsp;&nbsp; `max` : Giá trị lớn nhất trong cột.<br>

## Thay thế dữ liệu và thay đổi định dạng của dữ liệu

&#8226;&nbsp;&nbsp; Theo như thông tin có được ở trên thì tập dữ liệu không có giá trị thiếu (NaN) nên có thể bỏ qua phần bổ sung dữ liệu còn thiếu.<br>
&#8226;&nbsp;&nbsp; Tập dữ liệu có tất cả kiểu dữ liệu đều là dạng số, rất tiện cho máy học nên cũng không cần thiết thay đổi.<br>
&#8226;&nbsp;&nbsp; Bên cạnh đó ***target*** (`class`) có kiểu dữ `int` nên không cần thay đổi để phù hợp hơn.
